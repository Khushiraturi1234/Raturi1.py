```python
#3. write a fun ction to fiund the gcd of two number 
def gcd(a,b):
    n = min(a,b)
    while n:
        if a% n ==0 and b%n==0:
            break 
        n-=1
        return n 
gcd(12,45)
```

```python
#7. Write a function to generate Fibonacci series up to n terms.
def fibonacci_series(y):
    if y <= 0:
        return []
    elif y == 1:
        return [0]
    elif y == 2:
        return [0, 1]

    series = [0, 1]
    for i in range(2, y):
        series.append(series[-1] + series[-2])
    return series
print(fibonacci_series(10))
```

```python
#8. Write a function to count the number of vowels in a string.
def count_vowels(x):
    vowels = "aeiouAEIOU"
    return sum(1 for char in x if char in vowels)
print(count_vowels("Khushi Raturi"))
```

```python
#9 Write a function to find the maximum number in a list.
def max_no(n):
    if not n:
        return None 

    max_no = n[0]
    for no in n[1:]:
        if no > max_no:
            max_no = no
    return max_no
print(max_no([3, 7, 2, 9, 4]))
```

```python
#Write a function to find the sum of elements in a list
def sum(numbers):
    total=0 
    for num in numbers:
       total+=num
    return total 
print(sum([3, 7, 2, 9, 4]))
```

```python
# 13. Write a function to find the length of a list without using len()
def list_length(lst):
    count = 0
    for _ in lst:
        count += 1
    return count
print(list_length([]))       
```

```python
#16. Write a function to merge two dictionaries.
def merge_dicts(dict1, dict2):
    merged = dict1.copy()  
    merged.update(dict2)
    return merged
d1 = {'a': 1, 'b': 2}
d2 = {'b': 3, 'c': 4}
print(merge_dicts(d1, d2))  
```

```python
#18. Write a function to calculate the area of a circle given its radius.
import math

def circle_area(radius):
  
    if radius < 0:
        raise ValueError("Radius cannot be negative.")
    return math.pi * radius ** 2
print(circle_area(5)) 
```

```python
#19. Write a function to remove all punctuations from a string.  
import string

def remove_punctuation(text):
    return ''.join(char for char in text if char not in string.punctuation)
print(remove_punctuation("Hello, world! How's everything?"))  
```

```python
#20 20. Write a function to calculate LCM of two numbers.
def calculate_lcm(a, b):
    def gcd(x, y):
        while y:
            x, y = y, x % y
        return x

    return abs(a * b) // gcd(a, b)
print(calculate_lcm(12,18))
```
