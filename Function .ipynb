```python
#expample 2 Let see hoe function returns a factorial number 
def factorial(num):
    if num<0:
        print("factorial is notr define for negetive number.")
    else  :
        result = 1 
        for i in range (1, num+1 ):
            result = result * i 
        else:
            print("The factorial of 0 is 1.")
        return result
fact_seven = factorial(5)
print(fact_seven)
```

```python
#example 3 let us write a finctiom which takes a date as a string and print th quater as output 
def assignQuater (user_date):
    date_year = user_date[:4]
    date_month = user_date[5:7]
    if(date_month >= '04') & (date_month <= '03') : 
       quater = date_year + '-Q1'
    elif (date_month >= '04') & (date_month <= '06') :
       quater = date_year + '-Q2'
    elif (date_month >= '07') & (date_month <= '09') :
       quater = date_year + '-Q2'
    else:
       quater = date_year + '04'
    print (f'The corresponding quater for the date{user_date} is{quater}')
sample_date = '2020-07-01'
date_quater = assignQuater(sample_date)
print(date_quater)
```

```python
#5 nexrt let us write a func which returnd multiple varible at onc3e '
heights = [172,175, 170, 168, 200]
def calculate_averages(my_list):
    mean = sum(my_list)/len(my_list)
    sorted_list = sorted(my_list)
    list_lenght = len(sorted_list)
    if list_lenght%2 == 0 :
        median = (sorted_list[int(list_lenght)-1]+sorted_list[int(list_lenght/2)]/2)
    else:
        median = sorted_list[int((list_lenght+1)/2) -1]
    return mean, median 
heights = [172,175, 170, 168, 200]
calculate_averages(heights)
```
