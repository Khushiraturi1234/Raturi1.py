```python
#factorial of a no using loop 
```

```python
number = int(input("enter number to calculate its factorial:"))
factorial = 1
for i in range (1,number+1):
    factorial *= i
print(factorial )







```

```python
#Reverse a no using loop 
```

```python
num=int(input("enter the no"))
reversed_num = 0 
while num >0:
    digit = num%10
    reversed_num = reversed_num*10 + digit
    num//=10
print( reversed_num)
```
