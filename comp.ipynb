```python
#genrate squares of number from 1 to 10 
squares = {num:num**2 for num in range (1,11)}
print(squares)
```

```python
#extract even numbers from  a list 
l = [1,2,3,4,5,6]
even_number = [num**2 for num in l if num%2==0]
print(even_number)
```

```python
#Create a list of words longer than 3 charecters 
words = ["the", "cat", "elephant", "sun", "tree", "sky", "mountain"]

long_words = [word for word in words if len(word) > 3]

print("Words longer than 3 characters:", long_words)
```

```python
#List compehensionn - even number 
l = [1,2,3,4,5,6]
even_number = [num**2 for num in l if num%2==0]
print(even_number)
```

```python
#Tupple compehensionn - odd  number 
l = (1,2,3,4,5,6)
odd_squared_number = tuple(number**2 for number in l if number%2!=0)

print(odd_squared_number)
```

```python
#Tupple compehensionn - Reverse strings
words = ["hello", "world", "python", "tuple"]
reversed_tuple = tuple(word[::-1] for word in words)
print("Reversed strings in a tuple:", reversed_tuple)
```

```python
#set Compherision -vowels
text =  "Phython is my favouirite language "
vowels = {char.lower() for char in text  if char.lower() in 'aeiou'}

print("Vowels in the text:", vowels)
```

```python
#set comprehensions - Divisible by 5 
# Create a set of numbers divisible by 5 from 1 to 50
div_by_5 = {num for num in range(1, 51) if num % 5 == 0}

print("Numbers divisible by 5:", div_by_5)

```

```python
#Set Compreheinsion - unique words 
sentence = "Python is simple and powerful and Python is fun"
unique_words = {word.lower() for word in sentence.split()}
print("Unique words:", unique_words)
```

```python
#Dict omphersion - word lenghts 
sentence = "Dictionary comprehension is very useful"
word_lengths = {word: len(word) for word in sentence.split()}
print("Word lengths:", word_lengths)

```

```python
#dict comprehension - Swap key- value 
original_dict = {'a': 1, 'b': 2, 'c': 3}
swapped_dict = {value: key for key, value in original_dict .items()}
print(swapped_dict)

```

```python
#filter number divisible by both 3 and5 
numbers = [1, 5, 10, 15, 30, 45, 60, 70]
filtered = [num for num in numbers if num % 3 == 0 and num % 5 == 0]
print(filtered)
```
