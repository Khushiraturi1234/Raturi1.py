```python
#simple calculator using if else 
```

```python
p = int(input("enter first number"))
q = int(input("enter second number"))
oper = input("enter the type of opretor you want perform (+,-,%,*,):")

if (oper == '+' ):
    print (p+q)
elif(oper == '-' ):
    print (p-q)
elif (oper == '*'):
    print (p*q)
elif (oper == '%' ):
    print(p%q)
else:
    print ("invalid output")
    
```
