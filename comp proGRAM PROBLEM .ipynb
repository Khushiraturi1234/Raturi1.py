```python
#19 Temperatures in Celsius from Fahrenheit
#fahrenheit = [98.6, 102.2, 97.5]
fahrenheit = [98.6, 102.2, 97.5]
celsius = [round((f - 32) * 5 / 9, 2) for f in fahrenheit]
print(celsius)
#formula to convert freinate in celcius =  Celsius=5/9 ×(Fahrenheit−32)
```
