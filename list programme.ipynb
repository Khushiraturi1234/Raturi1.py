```python
#count even and odd no in list 
```

```python
s = [29, 50, 60, 100]
b= 0
c=0

for i in range (len(s)):
    if s [i]%2==0:
        b+=1
    else:
        c+=1
print (f"even numbers",b)
print (f"odd numbers",c)
```
