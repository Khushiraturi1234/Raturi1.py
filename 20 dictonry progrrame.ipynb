```python
#1. Create a Dictionary 
my_dict = {"name":"Alice","age":"20","city":"NewYork"}
print(my_dict)
```

```python
#2. Access  Elemnts 
print(my_dict["name"])
```

```python
#3.Add Key - Value Pair 
my_dict["email"]= "alice@example.com"
print(my_dict)
```

```python
#4. Update Value 
my_dict["age"] = 26 
print(my_dict)
```

```python
#5. Delete a Key 
del my_dict["city"]
print(my_dict)
```

```python
#6. Check if Key Exists
if "name" in my_dict:
 print("Key exists")
```

```python
#7. Get All Keys and Values
print(my_dict.keys())
print(my_dict.values())
```

```python
#8. Dictionary Length
print(len(my_dict))
```

```python
#9. Loop Through Dictionary
for key, value in my_dict.items():
    print(key,":",value)
```

```python
#10. Merge Two Dictionaries
dict1 = {"a": 1, "b": 2}
dict2 = {"c": 3}
dict1.update(dict2)
print(dict1)
```

```python
#11. Dictionary with List as Values
students = {
 "Alice": [85, 90, 95],
 "Bob": [78, 88, 84]
}
print(students["Alice"])

```

```python
#12. Count Frequency of Characters
string = "hello world"
freq = {}
for char in string:
 if char in freq:
     freq[char] += 1
 else:
    freq[char] = 1
print(freq)
```

```python
#13. Convert Two Lists into Dictionary
keys = ['a', 'b', 'c']
values = [1, 2, 3]
my_dict = dict(zip(keys, values))
print(my_dict)
```

```python
#14. Find Max Value in Dictionary
marks = {"John": 87, "Mike": 95, "Sara": 90}
topper = max(marks, key=marks.get)

print("Topper is", topper)
```

```python
#15.Remove Duplicate Values
my_dict = {'a': 1, 'b': 2, 'c': 1}
unique_dict = {}
for k, v in my_dict.items():
 if v not in unique_dict.values():
     unique_dict[k] = v
print(unique_dict)
```

```python
#16. Sort Dictionary by Key
d = {"banana": 3, "apple": 4, "pear": 1}
sorted_dict = dict(sorted(d.items()))
print(sorted_dict)
```

```python
#17. Sort Dictionary by Value
d = {"banana": 3, "apple": 4, "pear": 1}
sorted_dict = dict(sorted(d.items(), key=lambda item: item[1]))
print(sorted_dict)
```

```python
#18. Nested Dictionary
students = {
 "Alice": {"age": 20, "grade": "A"},
 "Bob": {"age": 22, "grade": "B"}
}
print(students["Alice"]["grade"])
```

```python
#19. Dictionary from User Input
n = int(input("Enter number of entries: "))
user_dict = {}
for _ in range(n):
 key = input("Enter key: ")
 value = input("Enter value: ")
 user_dict[key] = value
print(user_dict)
```

```python
#20. Dictionary Comprehension
squares = {x: x*x for x in range(1, 6)}
print(squares)
```
