```python
l = [1,2,3,6,11,56,8]
```

```python
squared_number = []
for no in l :
    squared_number.append(number**2)
print(f"The list of squared number is {squared_number}")
```

```python
squared_number = [number**2 for number in l]
print(squared_number)
```

```python
print(l)
```

```python
even_squared_number = [number**2 for number in l if number%2==0]
print(even_squared_number)
```

```python
odd_squared_number = [number**2 for number in l if number%2!=0]
print(odd_squared_number)
```

```python
new_numbers = [number**2 if number%2!=0 else number**3 for number in l ]
print(new_numbers)
```

```python
#using Loop 
even_squared_number =[]
for number in l :
    if number%2==0:
        even_squared_number.append(number**2)
        print(even_squared_number)
```

```python
#using list comprehension
even_squared_number = [number**2 for number in l if number%2==0 ]
print(even_squared_number)
```
